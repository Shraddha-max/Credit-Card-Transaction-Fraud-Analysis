```python
# Feature Engineering
```

```python
# Extract hour from transaction time
data['trans_date_trans_time'] = pd.to_datetime(data['trans_date_trans_time'])

data['transaction_hour'] = data['trans_date_trans_time'].dt.hour
```

```python
data['transaction_hour']
```

```python
# Create day of week
data['day_of_week'] = data['trans_date_trans_time'].dt.day_name()
```

```python
data['day_of_week']
```

```python
# Create month
data['month'] = data['trans_date_trans_time'].dt.month
data['month']
```

```python
# Create weekend indicator
data['is_weekend'] = data['trans_date_trans_time'].dt.dayofweek >= 5
data['is_weekend']
```

```python
# Create transaction amount categories
data['amount_category'] = pd.cut(
    data['amt'],
    bins=[0, 50, 200, 500, float('inf')],
    labels=['Low', 'Medium', 'High', 'Very High']
)
data['amount_category']
```

```python
# Create customer transaction count
# Is this a customer's first/rare transaction or part of normal activity?

data['customer_transaction_count'] = data.groupby('cc_num')['cc_num'].transform('count')
data['customer_transaction_count']
```

```python
# Create customer average spending
data['customer_avg_amount'] = (
    data.groupby('cc_num')['amt']
    .transform('mean')
)

data['customer_avg_amount'] 
```

```python
import numpy as np

lat1 = np.radians(data['lat'])
lon1 = np.radians(data['long'])

lat2 = np.radians(data['merch_lat'])
lon2 = np.radians(data['merch_long'])

dlat = lat2 - lat1
dlon = lon2 - lon1

a = np.sin(dlat / 2)**2 + np.cos(lat1) * np.cos(lat2) * np.sin(dlon / 2)**2

c = 2 * np.arcsin(np.sqrt(a))

data['distance_km'] = 6371 * c

data['distance_km']
```

```python
# Transaction Amount Histogram
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.hist(data['amt'], bins=50)

plt.xlabel('Transaction Amount')
plt.ylabel('Number of Transactions')
plt.title('Transaction Amount Distribution')

plt.show()
```

```python
# Fraud vs Non-Fraud Pie Chart

fraud_counts = data['is_fraud'].value_counts()

plt.figure(figsize=(6,6))

plt.pie(
    fraud_counts,
    labels=['Non-Fraud', 'Fraud'],
    autopct='%1.2f%%',
    startangle=90
)

plt.title('Fraud vs Non-Fraud Transactions')

plt.show()
```

```python
# Hourly Transaction Line Chart
hourly_transactions = (
    data.groupby(data['trans_date_trans_time'].dt.hour)
    .size()
)

plt.figure(figsize=(10,5))

plt.plot(
    hourly_transactions.index,
    hourly_transactions.values,
    marker='o'
)

plt.xlabel('Hour of Day')
plt.ylabel('Number of Transactions')
plt.title('Hourly Transaction Trends')

plt.xticks(range(24))

plt.show()
```

```python
# Merchant Category Bar Chart
category_transactions = (
    data.groupby('category')['amt']
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10,6))

plt.bar(
    category_transactions.index,
    category_transactions.values
)

plt.xlabel('Merchant Category')
plt.ylabel('Total Transaction Amount')
plt.title('Transaction Value by Merchant Category')

plt.xticks(rotation=45, ha='right')

plt.show()
```

```python
# Transaction Amount Box Plot

plt.figure(figsize=(8,5))

plt.boxplot(data['amt'])

plt.ylabel('Transaction Amount')
plt.title('Transaction Amount Box Plot')

plt.show()
```

```python
# Customer Spending Chart

customer_spending = (
    data.groupby('cc_num')['amt']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(10,6))

plt.bar(
    customer_spending.index.astype(str),
    customer_spending.values
)

plt.xlabel('Customer / Card Number')
plt.ylabel('Total Transaction Amount')
plt.title('Top 10 Customers by Total Spending')

plt.xticks(rotation=45, ha='right')

plt.show()
```

```python
# Scatter Plot Transaction Amount vs Customer Age

plt.figure(figsize=(10,6))

plt.scatter(
    data['Customer_Age'],
    data['amt'],
    alpha=0.5
)

plt.xlabel('Customer Age')
plt.ylabel('Transaction Amount')
plt.title('Customer Age vs Transaction Amount')

plt.show()
```

```python
# Correlation Heatmap

corr_data = data[
    [
        'amt',
        'Customer_Age',
        'Customer_Satisfaction_Score',
        'Loyalty_Points_Earned',
        'is_fraud'
    ]
]

correlation = corr_data.corr()

plt.figure(figsize=(8,6))

plt.imshow(correlation, cmap='coolwarm')

plt.colorbar()

plt.xticks(
    range(len(correlation.columns)),
    correlation.columns,
    rotation=45,
    ha='right'
)

plt.yticks(
    range(len(correlation.columns)),
    correlation.columns
)

plt.title('Correlation Heatmap')

plt.tight_layout()

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8,6))

sns.heatmap(
    correlation,
    annot=True,
    fmt='.2f',
    cmap='coolwarm'
)

plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()
```

```python
# Pair Plot
import seaborn as sns

pair_data = data[
    [
        'amt',
        'Customer_Age',
        'Customer_Satisfaction_Score',
        'Loyalty_Points_Earned',
        'is_fraud'
    ]
].sample(1000, random_state=42)

sns.pairplot(pair_data)

plt.show()
```

```python
# Fraud Distribution Count Plot

plt.figure(figsize=(7,5))

sns.countplot(
    data=data,
    x='is_fraud'
)

plt.xlabel('Fraud Status')
plt.ylabel('Number of Transactions')
plt.title('Fraud Distribution')

plt.xticks(
    [0, 1],
    ['Non-Fraud', 'Fraud']
)

plt.show()
```
